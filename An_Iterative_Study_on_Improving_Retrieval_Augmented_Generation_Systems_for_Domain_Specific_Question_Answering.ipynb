```python
import os
from langchain_community.document_loaders import PyPDFLoader

all_docs = []

for file in os.listdir('/content'):
    if file.endswith('.pdf'):
        loader = PyPDFLoader(f"/content/{file}")
        docs = loader.load()
        all_docs.extend(docs)

print("Total documents loaded:", len(all_docs))
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(
    chunk_size=300,
    chunk_overlap=50
)

chunks = splitter.split_documents(all_docs)

print("Total chunks:", len(chunks))
```

```python
from langchain_community.embeddings import HuggingFaceEmbeddings

embeddings = HuggingFaceEmbeddings()
```

```python
from langchain_community.vectorstores import FAISS

vectorstore = FAISS.from_documents(chunks, embeddings)
```

```python
query = "What is machine learning?"

results = vectorstore.similarity_search(query, k=3)
```

```python
for i, r in enumerate(results):
    print(f"Result {i+1}:")
    print(r.page_content)
    print("------")
```

```python
# ===============================
# STEP 1: Load FLAN-T5 Model
# ===============================
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

tokenizer = AutoTokenizer.from_pretrained("google/flan-t5-base")
model = AutoModelForSeq2SeqLM.from_pretrained("google/flan-t5-base")

def generate_answer(prompt):
    inputs = tokenizer(prompt, return_tensors="pt", truncation=True)
    outputs = model.generate(**inputs, max_length=200)
    return tokenizer.decode(outputs[0], skip_special_tokens=True)


# ===============================
# STEP 2: Define Queries (Same as Day 2)
# ===============================
queries = [
    "What is machine learning?",
    "What is supervised learning?",
    "Difference between supervised and unsupervised learning?",
    "What is backpropagation?"
]


# ===============================
# STEP 3: Run RAG Pipeline (Day 3 Improved)
# ===============================
for query in queries:

    # 🔍 Retrieval with MMR (diverse + relevant)
    results = vectorstore.max_marginal_relevance_search(query, k=5)

    # 🧩 Combine context
    context = " ".join([r.page_content for r in results])

    # 🧠 Prompt with instruction
    prompt = f"""
    Answer ONLY using the context below.

    If the answer is not present, say:
    "I don't know based on the provided data."

    Context:
    {context}

    Question:
    {query}

    Answer:
    """

    # 🤖 Generate answer
    answer = generate_answer(prompt)

    # 📊 Print output
    print("="*60)
    print("Question:", query)
    print("\nAnswer:", answer)
    print("="*60)
```
